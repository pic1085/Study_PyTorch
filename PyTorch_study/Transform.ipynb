### Transform
- 데이터가 항상 머신러닝 알고리즘 학습에 필요한 최종 처리가 된 형태로 제공되지 않음
- transform을 통해 데이터를 조작하고 학습에 적절하게 변형

```python
import torch
from torchvision import datasets
from torchvision.transforms import ToTensor, Lambda

ds = datasets.FashionMNIST(
    root="data",
    train=True,
    download=True,
    transform=ToTensor(),
    target_transform=Lambda(lambda y: torch.zeros(10, dtype=torch.float).scatter_(0, torch.tensor(y), value=1))
)
```

__ToTensor()__
- `ToTensor`는 PIL Image나 NumPy `ndarray`를 `FloatTensor`로 변환하고, 이미지의 픽셀의 크기 값을 [0., 1.] 범위로 비례하여 조정(scale) 함

__Lambda Transform__
- 정수를 one-hot 으로 부호화된 텐서로 바꾸는 함수를 정의
- 해당 함수는 크기 10짜리 0텐서를 만들고, `scatter_`를 호출하여 주어진 정답에 해당하는 인덱스에 `value=1`을 할당함

```python
target_transform = Lambda(lambda y: torch.zeros(10, dtype=torch.float).scatter_(0, torch.tensor(y), value=1)) # one-hot encoding
```
