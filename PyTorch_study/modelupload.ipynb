### 모델 저장 및 불러오기


```python
import torch
import torchvision.models as models
```

__모델 가중치 저장 후 불러오기__
- PyTorch 모델은 학습한 매개변수를 `state_dict`라고 불리는 내부 상태 사전에 저장됨(internal state dictionary).
- 이 상태 값들은 `torch.save` 메소드를 사용하여 저장할 수 있음

```python
model = models.vgg16(weights='IMAGENET1K_V1')
torch.save(model.state_dict(), 'model_weights.pth')
```

- 모델 가중치를 불러오기 위해, 동일한 모델의 인스턴스를 생성 후 `load_state_dict()` 메소드를 사용하여 매개변수 호출

```python
model = models.vgg16()
model.load_state_dict(torch.load('model_weights.pth'))
model.eval()
```

> 참고
- 추론을 하기 전 `model.eval()` 메소드를 호출하여 dropout과 batch normalization을 평가 모드(evaluation model)로 설정
    - 그렇지 않으면 일관성 없는 추론 결과가 생성됨

__모델의 형태를 포함하여 저장하고 불러오기__
- 모델의 가중치를 불러올 때, 신경망의 구조를 정의하기 위해 모델 클래스를 먼저 생성
- 구조를 모델과 함께 저장하고 싶으면 -> `model.state_dict()` 가 아닌 `model`을 저장 함수에 전달

```python
torch.save(model, 'model.pth')
```

```python
model = torch.load('model.pth')
```

> 참고
- 이 방식은 Python pickle 모듈을 사용하여 모델을 직렬화 하므로, 모델을 불러올 때 실제 클래스 정의를 적용해야 함
